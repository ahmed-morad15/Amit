```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('train.csv')
df.head()
```

```python
df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
df.head()
```

```python
def chk_types(df):
    dtypes =df.dtypes
    n_unique = df.nunique()
    return pd.DataFrame({"Dtypes":dtypes,"Num_uniques":n_unique}).T
```

```python
cols = ['Survived','Pclass','Parch','Sex','Embarked','SibSp'] 
 
df[cols] = df[cols].astype('category')
chk_types(df)
```

```python
null = df.isnull().sum()
ratio = round( (null /df.shape[0])*100 ,2).astype(str) +"%"
pd.DataFrame({"Null_sum": null, "Ratio %": ratio}).T
```

```python
col = 'Age'
Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
normal = df[(df[col] >= Q1 - 1.5*IQR) & (df[col] <= Q3 + 1.5*IQR)]

fig = go.Figure()
fig.add_trace(go.Scatter( y=normal[col], mode='markers', name='Normal Data', marker=dict(color='blue')))
fig.add_trace(go.Scatter( y=outliers[col], mode='markers', name='Outliers', marker=dict(color='red')))

fig.update_layout(title=f'<b>{col} Box Plot with Outliers Separated</b>', width=700, height=500)
fig.show()
```

```python
fig = sp.make_subplots(rows=2, cols=3, subplot_titles=cat_cols,
                       specs=[[{'type':'domain'}]*3, [{'type':'domain'}]*3])
 
for i, col in enumerate(cat_cols):
    row = i // 3 + 1
    col_pos = i % 3 + 1
   
    unique = df[col].value_counts()
    categories = unique.index
    count = unique.values
   
    fig.add_trace(go.Pie(labels=categories, values=count, name=col, hole=0.4), row=row, col=col_pos)
fig.update_layout(title_text='<b>Categorical Features Distribution</b>', width=900, height=600)
fig.show()
```

```python
import pandas as pd
import seaborn as sns
import plotly.express as px
 
# Load Titanic dataset
titanic = sns.load_dataset('titanic')
 
# Keep numeric columns and drop NaNs
data = titanic[['age', 'fare', 'sibsp', 'pclass']].dropna()
 
# Detect outliers using IQR
def detect_outliers(df, cols):
    outlier_mask = pd.Series(False, index=df.index)
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        outlier_mask |= (df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)
    return outlier_mask
 
data['outlier'] = detect_outliers(data, ['age', 'fare', 'sibsp'])
 
# Map outlier flag to string for better coloring
data['type'] = data['outlier'].map({True:'Outlier', False:'Normal'})
 
# 3D animated scatter
fig = px.scatter_3d(
    data,
    x='age', y='fare', z='sibsp',
    color='type',       # Normal vs Outlier
    symbol='type',
    animation_frame='pclass',  # Animate by passenger class
    size_max=10,
    title='Titanic 3D Scatter with Outliers Animated by Passenger Class'
)
 
fig.show()
 
```
