```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
df = pd.DataFrame(data = data['data'], columns=data.feature_names)
df['target'] = data['target']
df.head()
```

```python
df['target'].value_counts()
```

```python
x = df.iloc[:, :-1].values
y = df['target'].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def update_weights(x, y, weights, learning_rate, z):
    m = len(y)
    dw = (1 / m) * np.dot(x.T, (z - y))
    weights -= learning_rate * dw
    return weights
```

```python
def update_bias(y, z, bias, learning_rate):
    m = len(y)
    db = (1 / m) * np.sum(z - y)
    bias -= learning_rate * db
    return bias
```

```python
def perceptron(x, bias, weights):
    z = np.dot(x, weights) + bias
    y_pred = sigmoid(z)
    y_pred_cls = [1 if i > 0.5 else 0 for i in y_pred]
    return y_pred_cls
    
```

```python
def accuracy(y_actual, y_pred):
    acc = np.sum(y_actual == y_pred) / len(y_actual)
    return acc
```

```python
pred_test = perceptron(x_test, bias, weights)
print("Accuracy:", accuracy(y_test, pred_test))
```
